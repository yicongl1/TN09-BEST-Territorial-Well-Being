```python
# Import all libraries

import pandas as pd # data processing,
import numpy as np # linear algebra
import matplotlib.pyplot as plt # ploting the data
import seaborn as sns # ploting the data
import math # calculation
from scipy import stats #statistics
from collections import Counter

# Set up color blind friendly color palette
# The palette with grey:
cbPalette = ["#999999", "#E69F00", "#56B4E9", "#009E73", "#F0E442", "#0072B2", "#D55E00", "#CC79A7"]
# The palette with black:
cbbPalette = ["#000000", "#E69F00", "#56B4E9", "#009E73", "#F0E442", "#0072B2", "#D55E00", "#CC79A7"]

# sns.palplot(sns.color_palette(cbPalette))
# sns.palplot(sns.color_palette(cbbPalette))

sns.set_palette(cbPalette)
#sns.set_palette(cbbPalette)
```

```python
# Load the data

# The original notebook was run on Google Colab; data is now read from ./data/

DATA_PATH = 'data/OECD_BLI_all.csv'  # relative path in this repository
```

```python
encodings = ['utf-8', 'latin1', 'gbk']
for encoding in encodings:
    try:
        data = pd.read_csv(DATA_PATH, encoding=encoding)
        print(f"Successfully read the file with encoding: {encoding}")
        break
    except UnicodeDecodeError:
        print(f"Failed to read the file with encoding: {encoding}")
else:
    print("Failed to read the file with all tried encodings")
```

To study the correlation of indicators, it is necessary to  regroup the dataframe, organize the row by "Country" and the column by "Indicator" with value of "OBS_VALUE"

```python
#Filter original data
columns_of_interest = ["Country", "Indicator", "OBS_VALUE"]
filtered_data = data[columns_of_interest]

pivot_table = filtered_data.pivot_table(
    index="Country", columns="Indicator", values="OBS_VALUE"
)

# fill NA value with 0
pivot_table = pivot_table.fillna(0)
pivot_table.index
```

```python
# calculate correlation
correlation_matrix = pivot_table.corr()

# draw heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap of Indicators")
plt.show()
```

Now, let's try seprate these countries by region and make a comparison.

```python
# make a dictionary

# region_map = {
#     'Asia': ['Japan', 'Korea', 'Israel', 'Turkey'],
#     'Europe': ['Austria', 'Belgium', 'Czechia', 'Denmark', 'Estonia', 'Finland',
#                'France', 'Germany', 'Greece', 'Hungary', 'Iceland', 'Ireland',
#                'Italy', 'Latvia', 'Lithuania', 'Luxembourg', 'Netherlands', 'Poland',
#                'Portugal', 'Slovak Republic', 'Slovenia', 'Spain', 'Sweden',
#                'Switzerland', 'United Kingdom'],
#     'North America': ['Canada', 'United States', 'Mexico'],
#     'South America': ['Argentina', 'Brazil', 'Chile', 'Colombia'],
#     'Oceania': ['Australia', 'New Zealand'],
#     'Africa': ['South Africa'],
#     'Other': ['OECD - Total']
# }
region_map = {
  'High-income': ['Australia', 'Austria', 'Belgium', 'Canada', 'Costa Rica', 'Denmark', 'Estonia', 'Finland', 'France', 'Germany', 'Iceland', 'Ireland', 'Israel', 'Italy', 'Japan', 'Korea', 'Latvia', 'Lithuania', 'Luxembourg', 'Netherlands', 'New Zealand', 'Norway', 'OECD - Total', 'Poland', 'Portugal', 'Slovenia', 'Spain', 'Sweden', 'Switzerland', 'United Kingdom', 'United States'],
  'Middle-income': ['Brazil', 'Chile', 'Colombia', 'Czechia', 'Mexico', 'Russia', 'Slovak Republic', 'Turkey', 'South Africa']
}

# apply country to region
country_to_region = {}
for region, countries in region_map.items():
    for country in countries:
        country_to_region[country] = region

# add pivot_table
pivot_table['Region'] = pivot_table.index.map(country_to_region)

# create new table
region_pivot_tables = {}
for region in region_map.keys():
    region_pivot_tables[region] = pivot_table[pivot_table['Region'] == region].drop(columns='Region')

# draw heatmap for each region
for region, table in region_pivot_tables.items():
    correlation_matrix = table.corr()
    plt.figure(figsize=(12, 10))
    sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm")
    plt.title(f"Correlation Heatmap of Indicators - {region}")
    plt.show()
```

# 2.Conclusion

The heatmap of correlation shows the similar pattern of common sense, which adds to its credibility. For example, ‘life expectancy’ has 85% negative correlation with ‘dwellings without basic facilities’(measured by percentage of people with indoor flushing toilets in their home). That’s very reasonable, dwellings without basic facilities lead to poor hygiene conditions, thus making it easier for people to catch diseases. Another example, personal earnings have 80% positive correlation with ‘life satisfaction’, that’s obvious.
But it also shows some abnormal statistics, like ‘water quality’, it shows high correlation with ‘employment rate’ and ’life satisfaction’.
